<h3 style="text-align:right;color:inherit;margin-top:0;">هدف این زنگ</h3>
<p>در این جلسه قرار نیست همهٔ NumPy را یاد بگیریم. هدف این است که بتوانیم یک <strong>Array</strong> بسازیم و شکل آن را بخوانیم.</p>
<ul><li>تفاوت List و NumPy Array را ببینیم.</li><li><code>shape</code>، <code>ndim</code>، <code>size</code> و <code>dtype</code> را بخوانیم.</li><li>با <code>zeros</code>، <code>ones</code>، <code>arange</code> و <code>linspace</code> آرایه بسازیم.</li><li>با <code>astype</code> نوع داده را تغییر دهیم.</li></ul>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;"> چرا NumPy؟</h2>

</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>قبل از تعریف کتابخانه، رفتار List را امتحان می‌کنیم. حدس بزنید خروجی چیست:</p></div>

```python
a = [1, 2, 3, 4]

print(a * 2)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>ضرب List در ۲ یعنی تکرار کردن List، نه دو برابر کردن تک‌تک اعداد. حالا همان کار را با حلقه انجام دهید:</p></div>

```python
a = [1, 2, 3, 4]

b = []

for x in a:
    b.append(x * 2)

print(b)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])

print(a * 2)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;"> ساخت Array</h2>

</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>نوع اصلی داده در NumPy، <code>ndarray</code> است؛ یعنی آرایه‌ای با یک یا چند بُعد.</p></div>

```python
import numpy as np

a = np.array([10, 20, 30, 40])

print(a)
print(type(a))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>حالا این آرایهٔ دوبعدی را بسازید. آن را مثل جدول نمره‌ها در نظر بگیرید: هر سطر یک دانش‌آموز و هر ستون یک آزمون است.</p></div>

```python
scores = np.array([
    [18, 16, 20],
    [14, 19, 17],
    [20, 20, 18]
])

print(scores)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">مثال محوری جلسه</h3>
<p>از اینجا به بعد، <code>scores</code> را نگه می‌داریم تا ویژگی‌های مختلف Array را روی یک دادهٔ واقعی‌تر ببینیم.</p>
</div></div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">چهار ویژگی مهم Array</h2>

</div>

```python
a = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])

print("shape:", a.shape)
print("ndim:", a.ndim)
print("size:", a.size)
print("dtype:", a.dtype)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">راهنمای خواندن ویژگی‌ها</h3>
<p><code>shape</code> → اندازهٔ هر بُعد</p><p><code>ndim</code> → تعداد بُعدها</p><p><code>size</code> → تعداد کل عنصرها</p><p><code>dtype</code> → نوع دادهٔ عناصر</p>
</div></div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>در آرایهٔ بالا، <code>shape</code> برابر <code>(2, 4)</code> است؛ یعنی ۲ سطر و ۴ ستون.</p><p>دقت کنید <code>(5,)</code> یعنی یک آرایهٔ یک‌بعدی با ۵ عضو؛ ماتریس ۱×۵ نیست.</p></div>

```python
x = np.array([1, 2, 3])

print("shape:", x.shape)
print("ndim:", x.ndim)
print("size:", x.size)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>پیش‌بینی کنید dtype هر کدام چیست:</p></div>

```python
a = np.array([1, 2, 3])
b = np.array([1.5, 2.7, 3.1])
c = np.array([1, 2, 3.5])

print(a.dtype)
print(b.dtype)
print(c.dtype)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;"> ساخت Array بدون List</h2>

</div>

```python
print(np.zeros(5))
print(np.zeros((3, 4)))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p>در <code>np.zeros((3, 4))</code> آرگومان تابع یک Tuple به شکل <code>(3, 4)</code> است؛ یعنی ۳ سطر و ۴ ستون.</p></div>

```python
print(np.ones(5))
print(np.ones((2, 3)))
```

```python
print(np.arange(0, 10, 2))
print(np.arange(5))
```

```python
print(np.linspace(0, 10, 6))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">arange یا linspace؟</h3>
<p><code>arange</code> → فاصله یا <strong>step</strong> را تعیین می‌کنیم.</p><p><code>linspace</code> → تعداد نقاط را تعیین می‌کنیم.</p>
</div></div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;"> تغییر dtype با astype</h2>

</div>

```python
a = np.array([1, 2, 3])

b = a.astype(float)

print(a)
print(a.dtype)
print(b)
print(b.dtype)
```

```python
a = np.array([1.2, 2.9, 3.7])
b = a.astype(int)

print(b)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;"><p><code>astype(int)</code> گرد کردن نیست؛ بخش اعشاری را حذف می‌کند. بنابراین ۲٫۹ به ۲ تبدیل می‌شود.</p></div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;"> مزهٔ عملیات برداری</h2>

</div>

```python
scores = np.array([10, 12, 15, 18])

print(scores + 1)
print(scores * 2)
print(scores / 20)
print(scores > 14)
```
