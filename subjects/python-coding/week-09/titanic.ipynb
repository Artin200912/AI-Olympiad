```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load the public CSV directly; use the bundled copy if network access fails.
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
try:
    df = pd.read_csv(url)
    print("Source:", url)
except OSError as exc:
    data_path = Path("data/titanic_real.csv")
    if not data_path.exists():
        data_path = Path("../data/titanic_real.csv")
    if not data_path.exists():
        raise RuntimeError("Internet is unavailable and the bundled CSV is missing.") from exc
    print("Network unavailable; using the bundled CSV:", data_path)
    df = pd.read_csv(data_path)
if df.shape != (891, 12):
    raise ValueError("Unexpected Titanic data shape; check the source CSV.")
sns.set_theme(style="whitegrid", palette="colorblind")
print("Shape:", df.shape)
print("Missing values:")
print(df.isna().sum().to_string())
print(df.head(3).to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
sns.histplot(data=df, x="Age", bins=18, ax=axes[0])
axes[0].set_title("Age distribution (known ages)")
sns.countplot(data=df, x="Pclass", order=[1, 2, 3], ax=axes[1])
axes[1].set_title("Passengers by class")
survival_by_class = df.groupby("Pclass")["Survived"].mean()
survival_by_class.plot.bar(ax=axes[2], color="#0891b2")
axes[2].set_title("Survival rate by class")
axes[2].set_ylabel("Proportion survived")
plt.tight_layout()
plt.show()
print(survival_by_class.round(3).to_string())
```

$$
\overline{x}=\frac{\sum_{i=1}^{n}x_i}{n}
$$

$$
s=\sqrt{\frac{\sum_{i=1}^{n}(x_i-\overline{x})^2}{n-1}}

Z-score فاصله از میانگین را بر حسب انحراف معیار می‌سنجد. هیچ‌کدام حکم خودکار حذف داده نیست.</div></div>

Z-score فاصله از میانگین را بر حسب انحراف معیار می‌سنجد؛ روی توزیع بسیار چوله تفسیر آستانهٔ ثابت سخت‌تر است. پرچم‌گذاری آماری مرحلهٔ بررسی است و دلیل کافی برای حذف داده نیست.</div></div>

$$

```python
q1, q3 = df["Fare"].quantile([0.25, 0.75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
iqr_flag = (df["Fare"] < lower) | (df["Fare"] > upper)
fare_z = (df["Fare"] - df["Fare"].mean()) / df["Fare"].std()
print("IQR bounds:", round(lower, 2), round(upper, 2))
print("IQR flags:", int(iqr_flag.sum()))
print("|z| > 3 flags:", int(fare_z.abs().gt(3).sum()))
print(df.loc[iqr_flag, ["Pclass", "Fare"]].head().to_string(index=False))
```

<div dir="rtl" lang="fa" style="display:block;box-sizing:border-box;width:calc(100% - 20px);max-width:calc(100% - 20px);min-width:0;margin:10px 0 10px 10px;padding:12px 16px;border:1px solid rgba(148,163,184,0.35);border-right:5px solid #7c3aed;border-radius:8px;background:rgba(124,58,237,0.14);direction:rtl;text-align:right;white-space:normal;overflow-wrap:anywhere;overflow:hidden;line-height:1.9;color:inherit;"><strong style="display:block;margin-bottom:4px;">🧠 مبانی نظری و ریاضی: دقت و مبنای اکثریت</strong><div>هر پیش‌بینی را با برچسب واقعی همان سطر مقایسه می‌کنیم و نسبت تعداد درست‌ها را می‌گیریم. مبنای اکثریت برای همه یک برچسب می‌گوید، پس دقتش برابر سهم کلاس پرتعدادتر است. چون هر دو عدد را روی همین داده حساب کرده‌ایم، برای سنجش عملکرد روی دادهٔ تازه به ارزیابی جدا نیاز داریم.</div></div>

$$
\mathrm{Accuracy}=\frac{\text{تعداد پیش‌بینی‌های درست}}{\text{تعداد کل پیش‌بینی‌ها}}
$$

```python
prediction = df["Sex"].eq("female").astype(int)
majority_label = df["Survived"].mode().iloc[0]
baseline = pd.Series(majority_label, index=df.index)
accuracy = prediction.eq(df["Survived"]).mean()
baseline_accuracy = baseline.eq(df["Survived"]).mean()
print("Rule accuracy:", round(accuracy, 3))
print("Majority baseline:", round(baseline_accuracy, 3))
```
