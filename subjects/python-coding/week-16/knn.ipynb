### ۲-۱. اثبات متریک بودن Minkowski

نامنفی بودن، تقارن و صفر شدن فاصله فقط برای بردارهای برابر مستقیماً از تعریف نتیجه می‌شوند. بخش اصلی، اثبات شرط مثلثی است. برای مرتبهٔ بین یک و بی‌نهایت، توان مزدوج q را تعریف می‌کنیم:

</div>

</div>

$$
\boxed{1\lt p\lt \infty,\qquad q=\frac{p}{p-1}}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

بنابراین:

</div>

</div>

$$
\boxed{\frac{1}{p}+\frac{1}{q}=1}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

نامساوی هولدر:

</div>

</div>

$$
\sum_j|a_jb_j|\leq\|a\|_p\|b\|_q.
$$

### اثبات Hölder با Young

اگر یکی از نرم‌ها صفر باشد، حکم بدیهی است. در غیر این صورت از نامساوی Young برای اعداد نامنفی استفاده می‌کنیم:

</div>

</div>

$$
\boxed{AB\leq \frac{A^p}{p}+\frac{B^q}{q}}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

این رابطه از میانگین حسابی و هندسی وزن‌دار نتیجه می‌شود. مؤلفه‌های دو بردار را بر نرم متناظر تقسیم کرده و نامساوی را جمع می‌کنیم:

</div>

</div>

$$
\sum_j\frac{|a_j|}{\|a\|_p}\frac{|b_j|}{\|b\|_q}\leq\frac1p\sum_j\frac{|a_j|^p}{\|a\|_p^p}+\frac1q\sum_j\frac{|b_j|^q}{\|b\|_q^q}=1.
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

### اثبات Minkowski ـ گام اول

با ضرب طرفین در حاصل‌ضرب نرم‌ها هولدر ثابت می‌شود. اکنون تعریف می‌کنیم:

</div>

</div>

$$
\boxed{S=\|u+v\|_p}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

اگر S صفر باشد، حکم بدیهی است. در غیر این صورت نامساوی مثلثی قدرمطلق را به کار می‌گیریم:

</div>

</div>

$$
\begin{aligned} S^p&=\sum_j |u_j+v_j|\,|u_j+v_j|^{p-1}\\ &\leq\sum_j |u_j||u_j+v_j|^{p-1}+\sum_j |v_j||u_j+v_j|^{p-1}.\end{aligned}
$$

### گام دوم ـ اعمال Hölder

در هر دو جمله از هولدر استفاده می‌کنیم. چون رابطهٔ زیر برقرار است:

</div>

</div>

$$
\boxed{(p-1)q=p}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

نرم q عامل مشترک برابر مقدار زیر است:

</div>

</div>

$$
\boxed{S^{p-1}}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

سپس بر این مقدار مثبت تقسیم می‌کنیم:

</div>

</div>

$$
\begin{aligned}\left(\sum_j |u_j+v_j|^{(p-1)q}\right)^{1/q}&=S^{p-1},\\ S^p&\leq(\|u\|_p+\|v\|_p)S^{p-1},\\ \therefore\quad\|u+v\|_p&\leq\|u\|_p+\|v\|_p.\end{aligned}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

### گام سوم ـ حالت‌های مرزی

برای مرتبهٔ یک، نامساوی مثلثی قدرمطلق را روی مؤلفه‌ها جمع می‌کنیم. برای مرتبهٔ بی‌نهایت، بیشینهٔ قدرمطلق جمع از مجموع دو بیشینه بزرگ‌تر نیست. با قرار دادن روابط زیر، شرط مثلثی فاصله ثابت می‌شود:

</div>

</div>

$$
\boxed{p=1\quad\text{or}\quad p=\infty}
$$

$$
\boxed{u=x-y,\qquad v=y-z}
$$

$$
d_p(x,z)\leq d_p(x,y)+d_p(y,z).
$$

### چرا p کوچک‌تر از یک متریک نیست؟

در ابعاد حداقل دو، برای مرتبه‌های بین صفر و یک، سه نقطهٔ زیر را در نظر بگیرید:

</div>

</div>

$$
\boxed{0\lt p\lt 1,\qquad x=(1,0),\quad y=(0,0),\quad z=(0,1)}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

دو فاصلهٔ اول یک و فاصلهٔ مستقیم بیش از دو است:

</div>

</div>

$$
\boxed{2^{1/p}\gt 2}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #16a34a;background:rgba(34,197,94,0.075);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

پس شرط مثلثی نقض می‌شود. در یک بعد، فرمول برای هر p مثبت به قدرمطلق اختلاف تقلیل می‌یابد.

</div>

</div>

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #f59e0b;background:rgba(245,158,11,0.075);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">

### ۴. مقیاس ویژگی‌ها و پیش‌پردازش

اگر یک ویژگی در بازهٔ هزاران و دیگری در بازهٔ واحد باشد، ممکن است اولی فاصله را تحت سلطه قرار دهد. تغییر واحد یک ویژگی بدون تغییر معنا می‌تواند همسایهٔ انتخاب‌شده را عوض کند.

**StandardScaler**

</div>

</div>

$$
\boxed{\frac{x-\mu}{\sigma}}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

**MinMaxScaler** (برای بازهٔ ۰ تا ۱)

</div>

</div>

$$
\boxed{\frac{x-x_{\min}}{x_{\max}-x_{\min}}}
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">

**RobustScaler**

**Normalizer:** مقیاس‌بندی هر **نمونه** به نرم واحد، نه هر ویژگی.

آمار تبدیل‌های مبتنی بر ویژگی فقط روی Train فراگرفته می‌شود. StandardScaler لزوماً داده را نرمال‌توزیع نمی‌کند و RobustScaler نسبت به نقاط پرت مقاوم‌تر است، نه مصون.

</div>

```python
X_demo = np.array([[20, 1000], [40, 1100], [60, 1000], [80, 1200]], float)
q_demo = np.array([[21, 1065]], float)
raw = minkowski(X_demo, q_demo)
print('Raw distances:', raw, '| nearest:', raw.argmin())
for scaler in [StandardScaler(), MinMaxScaler(), RobustScaler()]:
    Xs = scaler.fit_transform(X_demo)
    qs = scaler.transform(q_demo)
    ds = minkowski(Xs, qs)
    print(type(scaler).__name__, ds, '| nearest:', ds.argmin())
print('Row norms after Normalizer:', np.linalg.norm(Normalizer().fit_transform(X_demo), axis=1))
```

d_{\mathrm{scaled}}(x,z)^2=\sum_j\frac{(x_j-z_j)^2}{\sigma_j^2}.
$$

$$

```python
def neighbor_weights(distances, mode='uniform'):
    distances = np.asarray(distances, float)
    if mode == 'uniform':
        return np.ones_like(distances)
    if mode != 'distance':
        raise ValueError('weights must be uniform or distance')
    zeros = distances == 0
    inverse = np.divide(1.0, distances, out=np.zeros_like(distances), where=~zeros)
    return np.where(zeros.any(axis=1, keepdims=True), zeros.astype(float), inverse)

class NumpyKNNClassifier:
    def __init__(self, k=5, p=2, weights='uniform'):
        self.k, self.p, self.weights = k, p, weights

    def fit(self, X, y):
        self.X_ = np.asarray(X, float).copy()
        self.y_ = np.asarray(y).copy()
        if self.X_.ndim != 2 or self.y_.shape != (len(self.X_),):
            raise ValueError('X must be 2D and y must be 1D with matching length')
        if not 1 <= self.k <= len(self.X_):
            raise ValueError('k must be between 1 and the training-set size')
        self.classes_ = np.unique(self.y_)
        return self

    def kneighbors(self, Q):
        D = pairwise_broadcast(Q, self.X_, self.p)
        indices = np.argsort(D, axis=1, kind='stable')[:, :self.k]
        distances = np.take_along_axis(D, indices, axis=1)
        return distances, indices

    def predict_proba(self, Q):
        distances, indices = self.kneighbors(Q)
        w = neighbor_weights(distances, self.weights)
        labels = self.y_[indices]
        # TODO: Sum the neighbor votes for each class.
        # Hint: Compare labels with each class and multiply by w.
        raise NotImplementedError("TODO: compute weighted class votes")
        return scores / scores.sum(axis=1, keepdims=True)

    def predict(self, Q):
        return self.classes_[self.predict_proba(Q).argmax(axis=1)]

knn = NumpyKNNClassifier(k=3).fit(X_small, y_small)
print('Neighbors:', knn.kneighbors(q_small))
print('Vote shares:', knn.predict_proba(q_small))
print('Prediction:', knn.predict(q_small))
```

### ۹. مرز تصمیم و ارتباط k با Bias–Variance

مرز تصمیم ناحیه‌ای است که پیش‌بینی کلاس عوض می‌شود. برای k برابر ۱ در هندسهٔ اقلیدسی، هر نمونه یک سلول Voronoi دارد؛ مرز کلاس تنها روی بخشی از مرز سلول‌ها قرار می‌گیرد.

- k کوچک: مدل محلی‌تر، معمولاً Bias کمتر و Variance بیشتر؛ حساس به نویز.
- k بزرگ: هموارسازی بیشتر، معمولاً Bias بیشتر و Variance کمتر؛ امکان از دست رفتن ساختار محلی.
- k برابر n و رأی یکنواخت: پیش‌بینی همهٔ نقاط همان کلاس اکثریت مجموعهٔ آموزش است.

این‌ها روندهای معمول‌اند، نه قضیه‌ای دربارهٔ هر مجموعه‌داده. بهترین k را باید با دادهٔ اعتبارسنجی انتخاب کرد، نه از روی دقت آموزش.

<h3 style="margin:0 0 8px 0;text-align:right;color:inherit;">۱۲. انتخاب مدل با Pipeline و GridSearchCV</h3>
<p style="margin:6px 0;text-align:right;">برای جلوگیری از Data Leakage، ابتدا مجموعهٔ Test را کنار می‌گذاریم. سپس برای هر ترکیب Hyperparameter، پیش‌پردازش و مدل را داخل هر fold و فقط روی قسمت آموزشی همان fold fit می‌کنیم. Pipeline این فرایند را مدیریت می‌کند.</p>
<p style="margin:6px 0;text-align:right;">در این مثال پارامترهای <code>n_neighbors</code>، <code>p</code> (نوع فاصله)، <code>weights</code> و نوع Scaler با اعتبارسنجی متقاطع طبقه‌بندی‌شده انتخاب می‌شوند. معیار انتخاب Macro-F1 است:</p>
</div>
</div>

$$
F_{1,c}=\frac{2P_cR_c}{P_c+R_c},\qquad F_{1,\mathrm{macro}}=\frac1C\sum_{c=1}^C F_{1,c}.
$$

<div dir="rtl" lang="fa" style="font-family:'XB Niloofar','XB Niloofar Regular','B Nazanin','Vazirmatn','Tahoma',sans-serif;font-size:19px;direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(14,165,233,0.07);padding:14px 18px;border-radius:8px;margin:12px 0;color:inherit;">
<p style="margin:6px 0;text-align:right;"><code>GridSearchCV</code> با <code>refit=True</code> پس از انتخاب تنظیمات، بهترین Pipeline را روی کل دادهٔ Train بازآموزی می‌کند. Test در انتخاب تنظیمات شرکت نمی‌کند. میانگین CV معیار انتخاب است، نه برآورد کاملاً بی‌طرف پس از جست‌وجوی متعدد.</p>
</div>
</div>

```python
dataset = load_breast_cancer()
X_case, y_case = dataset.data, dataset.target
X_train, X_test, y_train, y_test = train_test_split(
    X_case, y_case, test_size=.25, stratify=y_case, random_state=SEED)
print('Train:', X_train.shape, '| Test:', X_test.shape)
print('Training class counts:', np.bincount(y_train))

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
knn_pipe = Pipeline([('scale', StandardScaler()), ('knn', KNeighborsClassifier(metric='minkowski'))])
param_grid = {
    'scale': ['passthrough', StandardScaler(), RobustScaler()],
    'knn__n_neighbors': [3, 7, 15],
    'knn__p': [1, 2],
    'knn__weights': ['uniform', 'distance'],
}
search_knn = GridSearchCV(knn_pipe, param_grid, cv=cv, scoring='f1_macro', n_jobs=1, refit=True)
search_knn.fit(X_train, y_train)
print('Best parameters:', search_knn.best_params_)
print('Best CV macro-F1:', round(search_knn.best_score_, 4))
results = search_knn.cv_results_
for i in np.argsort(results['rank_test_score'], kind='stable')[:5]:
    print(f"{results['mean_test_score'][i]:.4f} +/- {results['std_test_score'][i]:.4f}", results['params'][i])
```

### ۱۳. ارزیابی نهایی و مقایسه با Logistic Regression

برای مقایسهٔ منصفانه، هر مدل روی یک تقسیم مشترک Train/Test و با یک سیاست انتخاب مدل (CV و Macro-F1) ارزیابی می‌شود. برای Logistic Regression نیز مقدار C با CV انتخاب می‌شود.

- Accuracy: نسبت کل پیش‌بینی‌های درست.
- Balanced Accuracy: میانگین Recall کلاس‌ها.
- Macro-F1: میانگین F1 کلاس‌ها.
- ROC-AUC: کیفیت رتبه‌بندی امتیاز کلاس مثبت (در این مثال کلاس ۱، یعنی benign). برای KNN از نسبت رأی وزن‌دار به‌عنوان امتیاز استفاده می‌کنیم.

نتایج مربوط به همین داده و تقسیم‌اند؛ بهترین الگوریتم عمومی را تعیین نمی‌کنند. با دادهٔ نامتوازن، Accuracy ممکن است وضعیت کلاس کوچک‌تر را پنهان کند. ماتریس اغتشاش و معیارهای هر کلاس برای تشخیص این وضعیت مفیدند. اگر پس از دیدن Test تنظیمات مدل را تغییر دهیم، دیگر آن Test مستقل نیست.

</div>

</div>

```python
log_pipe = Pipeline([('scale', StandardScaler()),
                     ('lr', LogisticRegression(max_iter=3000, random_state=SEED))])
search_lr = GridSearchCV(log_pipe, {'lr__C': [.01, .1, 1, 10, 100]},
                         cv=cv, scoring='f1_macro', n_jobs=1, refit=True)
search_lr.fit(X_train, y_train)
print('Logistic Regression best params:', search_lr.best_params_)
models = {'KNN': search_knn.best_estimator_, 'Logistic Regression': search_lr.best_estimator_}
print(f"{'Model':<22} {'Accuracy':>10} {'Bal.Acc':>10} {'Macro-F1':>10} {'ROC-AUC':>10}")
fig, axes = plt.subplots(1, 2, figsize=(9, 4), constrained_layout=True)
for (name, model), ax in zip(models.items(), axes):
    pred = model.predict(X_test)
    pos = np.flatnonzero(model.classes_ == 1)[0]
    score = model.predict_proba(X_test)[:, pos]
    metrics = [accuracy_score(y_test, pred), balanced_accuracy_score(y_test, pred),
               f1_score(y_test, pred, average='macro'), roc_auc_score(y_test, score)]
    print(f'{name:<22}' + ''.join(f'{v:10.4f}' for v in metrics))
    ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=dataset.target_names,
                                           ax=ax, colorbar=False)
    ax.set_title(name)
plt.show()
print('KNN class-wise results:')
print(classification_report(y_test, models['KNN'].predict(X_test), target_names=dataset.target_names))
```

### ۱۳-۱. آزمایش مستقل KNN روی دادهٔ نامتوازن ۹۵ به ۵

دادهٔ مصنوعی دودویی با ۲۴۰۰ نمونه، دو ویژگی و نسبت ۹۵٪ کلاس صفر به ۵٪ کلاس یک تولید می‌کنیم. تقسیم Train/Test به‌صورت stratified است و کلاس یک اقلیت محسوب می‌شود. مدل Dummy همیشه اکثریت را پیش‌بینی می‌کند. سه مقدار از پیش مشخص k برابر ۱، ۱۵ و ۵۱ و رأی وزن‌دار با k برابر ۱۵ را مقایسه می‌کنیم.

این آزمایش برای نمایش رفتار مدل‌هاست، نه انتخاب بهترین تنظیمات با Test. برای انتخاب واقعی k باید از اعتبارسنجی روی Train استفاده کرد. مدل اکثریت Accuracy برابر ۰٫۹۵ و Recall اقلیت صفر دارد؛ Balanced Accuracy آن ۰٫۵ و Macro-F1 تقریباً ۰٫۴۸۷ است.

</div>

</div>

$$
\mathrm{Balanced\ Accuracy}=\frac{R_0+R_1}{2},\qquad F_{1,1}=\frac{2P_1R_1}{P_1+R_1}.
$$

```python
from sklearn.datasets import make_classification
from sklearn.dummy import DummyClassifier
from sklearn.metrics import precision_score, recall_score
X_imb, y_imb = make_classification(n_samples=2400, n_features=2,
    n_informative=2, n_redundant=0, n_clusters_per_class=1,
    weights=[.95,.05], flip_y=0, class_sep=.8, random_state=SEED)
Xi_train, Xi_test, yi_train, yi_test = train_test_split(
    X_imb, y_imb, test_size=.30, stratify=y_imb, random_state=SEED)
print('Train counts:', np.bincount(yi_train), '| Test counts:', np.bincount(yi_test))
imb_models = {'Majority': DummyClassifier(strategy='most_frequent')}
for ki, wi in [(1,'uniform'), (15,'uniform'), (51,'uniform'), (15,'distance')]:
    imb_models[f'k={ki}, {wi}'] = Pipeline([('scale', StandardScaler()),
        ('knn', KNeighborsClassifier(n_neighbors=ki, weights=wi))])
imb_names, imb_results = [], []
fig, axes = plt.subplots(2, 3, figsize=(12,7), constrained_layout=True)
axes = axes.ravel()
for (name, estimator), ax in zip(imb_models.items(), axes):
    estimator.fit(Xi_train, yi_train)
    pred_i = estimator.predict(Xi_test)
    score_i = estimator.predict_proba(Xi_test)[:, list(estimator.classes_).index(1)]
    values = [accuracy_score(yi_test,pred_i), balanced_accuracy_score(yi_test,pred_i),
        precision_score(yi_test,pred_i,zero_division=0), recall_score(yi_test,pred_i,zero_division=0),
        f1_score(yi_test,pred_i,zero_division=0), f1_score(yi_test,pred_i,average='macro',zero_division=0),
        roc_auc_score(yi_test,score_i)]
    imb_names.append(name); imb_results.append(values)
    ConfusionMatrixDisplay.from_predictions(yi_test,pred_i,labels=[0,1],ax=ax,colorbar=False,cmap='Blues')
    ax.set_title(name)
axes[-1].axis('off'); plt.show()
print(f"{'Model':<19} {'Acc':>7} {'BalAcc':>8} {'Prec1':>8} {'Rec1':>8} {'F1_1':>8} {'MacroF1':>8} {'AUC':>8}")
for name, values in zip(imb_names, imb_results):
    print(f'{name:<19}' + ''.join(f'{v:8.3f}' for v in values))
imb_arr = np.array(imb_results); xpos = np.arange(len(imb_names))
fig, ax = plt.subplots(figsize=(10,4), constrained_layout=True)
for j, (col,title) in enumerate([(0,'Accuracy'),(3,'Minority recall'),(5,'Macro-F1')]):
    ax.bar(xpos+(j-1)*.24,imb_arr[:,col],width=.24,label=title)
ax.set(xticks=xpos,xticklabels=imb_names,ylim=(0,1.05),ylabel='Score',title='Accuracy can hide minority-class errors')
ax.legend(); plt.show()
assert np.isclose(imb_arr[0,0],.95) and imb_arr[0,3]==0
```
