<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(37,99,235,0.07);padding:14px 18px;border-radius:8px;margin:10px 0;color:inherit;">
<h3 style="margin:0 0 8px 0;text-align:right;color:inherit;">pathlib چیست؟ Path چیست؟</h3>
<p><strong>pathlib</strong> یک ماژول استاندارد داخل Python برای کار با مسیر فایل‌ها و پوشه‌هاست؛ بنابراین لازم نیست مسیرها را همیشه با رشته و علامت‌های <code>/</code> یا <code>\</code> به‌صورت دستی کنار هم بچسبانیم.</p><p>از کلاس <strong>Path</strong> برای ساختن یک شیء مسیر استفاده می‌کنیم: <code>data_folder = Path("data")</code> و <code>students_file = data_folder / "students.csv"</code>.</p><p><code>data_folder</code> و <code>students_file</code> خودِ پوشه و فایل نیستند؛ آن‌ها شیءهایی هستند که <strong>آدرس</strong> را نمایش می‌دهند و ابزارهایی برای بررسی و مدیریت آن مسیر دارند. ساختن <code>Path("data")</code> به‌تنهایی پوشه را ایجاد نمی‌کند و فایل را هم باز نمی‌کند.</p><ul><li><strong>Relative Path:</strong> مثل <code>data/students.csv</code>؛ نسبت به پوشه‌ای که برنامه از آن اجرا شده است.</li><li><strong>Absolute Path:</strong> مسیر کامل فایل در سیستم.</li><li><strong>exists:</strong> بررسی می‌کند مسیر واقعاً وجود دارد یا نه.</li><li><strong>is_file / is_dir:</strong> مشخص می‌کند مسیر فایل است یا پوشه.</li><li><strong>name / parent / suffix:</strong> نام، پوشهٔ والد و پسوند مسیر را می‌دهد.</li></ul><p>عملگر <code>/</code> در <code>data_folder / "students.csv"</code> برای چسباندن بخش‌های مسیر استفاده شده است؛ این با تقسیم عددی فرق دارد و توسط <code>Path</code> مدیریت می‌شود.</p>
</div></div>

```python
from pathlib import Path

data_folder = Path("data")
students_file = data_folder / "students.csv"

print("Folder exists:", data_folder.exists())
print("File exists:", students_file.exists())
print("Is a directory:", data_folder.is_dir())
print("Is a file:", students_file.is_file())
print("File name:", students_file.name)
print("Parent:", students_file.parent)
csv_files = list(data_folder.glob("*.csv"))
print("CSV files:", csv_files)
for file in csv_files:
    print(" -", file.name, file.suffix)

all_csv_files = list(data_folder.glob("**/*.csv"))
print("CSV files including subfolders:", all_csv_files)
```

<h3 style="margin:0 0 8px 0;text-align:right;color:inherit;">glob را قدم‌به‌قدم بخوان</h3>
<p><code>data_folder.glob("*.csv")</code> را این‌طور بخوان: «داخل مسیر data_folder بگرد و هر مسیر مستقیمی را که نامش با هر چیزی شروع می‌شود و به <code>.csv</code> ختم می‌شود پیدا کن.»</p><p><code>*</code> یک الگوی جایگزین است؛ یعنی هر تعداد حرف یا عدد. پس <code>*.csv</code> با <code>students.csv</code> و <code>scores.csv</code> جور می‌شود، اما با فایل <code>students.txt</code> جور نمی‌شود.</p><p>خروجی <code>glob</code> معمولاً چند شیء <code>Path</code> است. برای دیدن آن‌ها می‌توانیم <code>list(...)</code> بنویسیم و برای چاپ فقط نام فایل از <code>file.name</code> استفاده کنیم. اگر بخواهیم زیرپوشه‌ها هم بررسی شوند، از <code>**/*.csv</code> یا <code>rglob("*.csv")</code> استفاده می‌کنیم.</p><table style="width:100%;border-collapse:collapse;text-align:right;"><tr><th style="padding:6px;border-bottom:1px solid #94a3b8;">الگو</th><th style="padding:6px;border-bottom:1px solid #94a3b8;">معنی</th></tr><tr><td style="padding:6px;"><code>*.csv</code></td><td style="padding:6px;">فایل‌های CSV در همین پوشه</td></tr><tr><td style="padding:6px;"><code>**/*.csv</code></td><td style="padding:6px;">فایل‌های CSV در این پوشه و زیرپوشه‌ها</td></tr><tr><td style="padding:6px;"><code>*.txt</code></td><td style="padding:6px;">فایل‌های متنی با پسوند txt</td></tr></table>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(37,99,235,0.07);padding:14px 18px;border-radius:8px;margin:10px 0;color:inherit;">
<h3 style="margin:0 0 8px 0;text-align:right;color:inherit;">متن، بایت و UTF-8</h3>
<p>کامپیوتر در پایین‌ترین سطح با <strong>بایت‌ها</strong> کار می‌کند، اما ما با حروف و کلمه‌ها کار می‌کنیم. Encoding روشی است که مشخص می‌کند هر متن چگونه به بایت تبدیل شود.</p>
<ul><li><strong>Encoding:</strong> تبدیل متن به بایت برای ذخیره‌سازی یا ارسال.</li><li><strong>Decoding:</strong> تبدیل بایت‌ها به متن قابل‌خواندن.</li><li><strong>Unicode:</strong> استانداردی برای نمایش حروف زبان‌های مختلف.</li><li><strong>UTF-8:</strong> یکی از رایج‌ترین روش‌های ذخیره‌کردن Unicode.</li></ul>
<p>هنگام نوشتن و خواندن یک فایل باید Encoding هماهنگ باشد. اگر فایل را با UTF-8 بنویسیم ولی با روش دیگری بخوانیم، ممکن است متن ناخوانا شود یا خطایی مثل <code>UnicodeDecodeError</code> رخ دهد.</p>
<pre style="direction:ltr;text-align:left;color:inherit;">text → encode → bytes → file
file → bytes → decode → text</pre>
<p>بهتر است Encoding را صریح بنویسیم و به مقدار پیش‌فرض سیستم‌عامل اعتماد نکنیم:</p>
<pre style="direction:ltr;text-align:left;color:inherit;">open(file, "r", encoding="utf-8")
open(file, "w", encoding="utf-8")</pre>
</div></div>

```python
text = "علی"
encoded_text = text.encode("utf-8")
decoded_text = encoded_text.decode("utf-8")

print("Text:", text)
print("Encoded bytes:", encoded_text)
print("Decoded text:", decoded_text)
print(type(encoded_text))
print(type(decoded_text))
```

<p>CSV قالبی ساده برای دادهٔ جدولی است. DictReader نام ستون‌ها را به کلیدهای Dictionary تبدیل می‌کند؛ اما مقدارها هنوز رشته‌اند و باید خودمان نوعشان را تبدیل کنیم.</p>

```python
csv_file = Path("data/students.csv")
rows = []
with open(csv_file, "r", encoding="utf-8", newline="") as file:
    reader = csv.DictReader(file)
    for row in reader:
        rows.append(row)
print(rows)
print("Type of age:", type(rows[0]["age"]))
```
