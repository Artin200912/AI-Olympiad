## 1. Why functions?

A function gives a name to a reusable process.

Functions help us:

- avoid repeated code;
- make programs easier to read;
- divide a large problem into smaller parts;
- test one part independently;
- reuse code in another program.

### Repeated code

The following program calculates two averages by repeating the same logic.

```python
group_a = [18, 14, 20]
group_b = [11, 16, 19]

total_a = 0
for score in group_a:
    total_a += score
average_a = total_a / len(group_a)

total_b = 0
for score in group_b:
    total_b += score
average_b = total_b / len(group_b)

print(average_a)
print(average_b)
```

### Reusable function

The repeated process can be moved into one function.

```python
def calculate_average(numbers):
    total = 0
    for number in numbers:
        total += number
    return total / len(numbers)

print(calculate_average(group_a))
print(calculate_average(group_b))
```

- `print` displays information.
- `return` sends a value back to the caller.
- A returned value can be stored, compared, tested, or passed to another function.

```python
def show_square(number):
    print(number ** 2)

def get_square(number):
    return number ** 2

value_from_print = show_square(5)
value_from_return = get_square(5)

print("Stored from print:", value_from_print)
print("Stored from return:", value_from_return)
```

## 6. Early return

A function can stop as soon as the answer is known.

```python
def first_negative_index(numbers):
    for index, number in enumerate(numbers):
        if number < 0:
            return index
    return None

print(first_negative_index([4, 8, -2, 7]))
print(first_negative_index([4, 8, 2, 7]))
```

## 8. Mutable arguments

Lists are mutable. A function may change the same list object that was passed to it.

Compare an in-place function with a function that returns a new list.

```python
def add_bonus_in_place(scores, bonus):
    for index in range(len(scores)):
        scores[index] += bonus

def scores_with_bonus(scores, bonus):
    result = []
    for score in scores:
        result.append(score + bonus)
    return result

original_scores = [12, 15, 18]
copied_scores = original_scores.copy()

add_bonus_in_place(copied_scores, 1)
new_scores = scores_with_bonus(original_scores, 1)

print("Original:", original_scores)
print("Changed copy:", copied_scores)
print("New list:", new_scores)
```

## 9. Breaking a problem into smaller functions

The following program analyses student records.

Each function has one clear responsibility.

```python
students = [
    {"name": "Ali", "score": 18},
    {"name": "Hossein", "score": 9},
    {"name": "Aria", "score": 14},
    {"name": "Kasra", "score": 20},
    {"name": "Reza", "score": 16},
]

def extract_scores(records):
    return [record["score"] for record in records]

def safe_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

def top_student(records):
    if not records:
        return None

    best = records[0]
    for record in records[1:]:
        if record["score"] > best["score"]:
            best = record
    return best

def count_passing(records, pass_mark=10):
    count = 0
    for record in records:
        if record["score"] >= pass_mark:
            count += 1
    return count

def build_report(records, pass_mark=10):
    scores = extract_scores(records)
    return {
        "average": safe_average(scores),
        "top_student": top_student(records),
        "passing_count": count_passing(records, pass_mark),
    }

report = build_report(students)
report
```

## 10. Docstrings

A docstring briefly explains what a function does.

```python
def clamp(value, lower, upper):
    """Return value limited to the inclusive interval [lower, upper]."""
    if lower > upper:
        raise ValueError("lower must not be greater than upper")
    if value < lower:
        return lower
    if value > upper:
        return upper
    return value

print(clamp.__doc__)
```

## 2. Countdown

Predict the output before running the cell.

```python
def countdown(number):
    if number == 0:
        print("Go!")
        return

    print(number)
    countdown(number - 1)

countdown(4)
```

## 3. Factorial

Factorial follows this recurrence:

$$n! = n \times (n-1)!$$

Base case:

$$0! = 1$$

```python
def factorial(number):
    if number < 0:
        raise ValueError("factorial is not defined for negative integers")
    if number == 0:
        return 1
    return number * factorial(number - 1)

print(factorial(5))
```

## 4. Recursive sum

```python
def recursive_sum(values, index=0):
    if index == len(values):
        return 0
    return values[index] + recursive_sum(values, index + 1)

print(recursive_sum([4, 7, 2, 5]))
```

## 5. Recursive string reversal

```python
def reverse_string(text):
    if len(text) <= 1:
        return text
    return reverse_string(text[1:]) + text[0]

print(reverse_string("python"))
```

## 6. Recursive palindrome check

```python
def is_palindrome(text):
    if len(text) <= 1:
        return True
    if text[0] != text[-1]:
        return False
    return is_palindrome(text[1:-1])

print(is_palindrome("level"))
print(is_palindrome("coding"))
```

## 7. Recursive maximum

```python
def recursive_max(values, index=0):
    if not values:
        raise ValueError("values must not be empty")

    if index == len(values) - 1:
        return values[index]

    maximum_of_rest = recursive_max(values, index + 1)
    return max(values[index], maximum_of_rest)

print(recursive_max([8, 3, 11, 5, 9]))
```

## 9. Pascal's triangle

For zero-based row `n` and position `k`:

$$P(n,k) =
\begin{cases}
1, & k=0 \text{ or } k=n \\
P(n-1,k-1) + P(n-1,k), & \text{otherwise}
\end{cases}$$

```python
def pascal_value(row, position):
    if position < 0 or position > row:
        return 0
    if position == 0 or position == row:
        return 1
    return (
        pascal_value(row - 1, position - 1)
        + pascal_value(row - 1, position)
    )

def pascal_row(row):
    return [pascal_value(row, position) for position in range(row + 1)]

for row in range(6):
    print(pascal_row(row))
```

## 10. Tower of Hanoi

Rules:

- move one disk at a time;
- move only the top disk;
- never place a larger disk on a smaller disk.

Recursive plan:

1. move `n - 1` disks from source to helper;
2. move the largest disk from source to destination;
3. move `n - 1` disks from helper to destination.

```python
def hanoi_moves(number_of_disks, source="A", destination="C", helper="B"):
    if number_of_disks == 0:
        return []

    moves = hanoi_moves(
        number_of_disks - 1,
        source,
        helper,
        destination,
    )
    moves.append((source, destination))
    moves.extend(
        hanoi_moves(
            number_of_disks - 1,
            helper,
            destination,
            source,
        )
    )
    return moves

for step, move in enumerate(hanoi_moves(3), start=1):
    print(step, move)
```
