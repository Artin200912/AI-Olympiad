# Part 1 — Representing the Board

We do not need an `8 × 8` matrix.

We place exactly one queen in each row. Therefore, a single list is enough.

The list index represents the row.  
The value at that index represents the queen's column.

Example:

```text
positions = [1, 3, 0, 2]
```

means:

- row `0` has a queen in column `1`;
- row `1` has a queen in column `3`;
- row `2` has a queen in column `0`;
- row `3` has a queen in column `2`.

````python
example_positions = [1, 3, 0, 2]

for row, column in enumerate(example_positions):
    print(f"Row {row}: queen in column {column}")
````

## Board display helper

This function is already complete. It displays a board using `Q` for a queen and `.` for an empty square.

````python
def board_as_text(positions, size=None):
    if size is None:
        size = len(positions)

    rows = []

    for row in range(size):
        cells = []

        for column in range(size):
            if row < len(positions) and positions[row] == column:
                cells.append("Q")
            else:
                cells.append(".")

        rows.append(" ".join(cells))

    return "\n".join(rows)


def print_board(positions, size=None):
    print(board_as_text(positions, size))


print_board(example_positions)
````

# Part 3 — Start with Four Queens

Before solving Eight Queens, solve the smaller `4 × 4` version.

A useful strategy is:

1. place a queen in the current row;
2. move to the next row;
3. if no column works, undo the last choice;
4. try another column.

This is called **backtracking**.

## Backtracking idea

At each row:

- choose a safe column;
- add it to `positions`;
- recursively solve the next row;
- remove the choice when returning.

The removal step is the **backtrack**.
