<h2 style="text-align:right;color:inherit;">بردار، ماتریس و Scalar</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>Scalar یک عدد، Vector یک فهرست جهت‌دار از عددها و Matrix یک جدول عددی است. در NumPy معمولاً Vector را با Array یک‌بعدی و Matrix را با Array دوبعدی نمایش می‌دهیم.</p>
</div>

```python
import numpy as np

scalar = 3
vector = np.array([2, 5, 1])
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("scalar:", scalar)
print("vector shape:", vector.shape)
print("matrix shape:", matrix.shape)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">شکل بردار</h3>
<p>بردار <code>[2, 5, 1]</code> در NumPy شکل <code>(3,)</code> دارد. این بردار نه سطری است و نه ستونی؛ برای ساختن ستون باید آن را به شکل <code>(3, 1)</code> تغییر دهیم.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">عملیات پایه روی بردارها</h2>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>دو بردار هم‌اندازه را عنصر‌به‌عنصر جمع یا کم می‌کنیم. ضرب در Scalar نیز تک‌تک مؤلفه‌ها را تغییر می‌دهد.</p>

```python
u = np.array([1, 2, 3])
v = np.array([4, 0, -1])

print("u + v:", u + v)
print("u - v:", u - v)
print("3u:", 3 * u)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">ضرب داخلی یا Dot Product</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>برای ضرب داخلی، مؤلفه‌های هم‌مکان را ضرب و سپس حاصل‌ها را جمع می‌کنیم:</p><p><code>[u1,u2,u3] · [v1,v2,v3] = u1v1 + u2v2 + u3v3</code></p>
</div>

```python
u = np.array([1, 2, 3])
v = np.array([4, 0, -1])

manual_dot = 1*4 + 2*0 + 3*(-1)
numpy_dot = u @ v

print("manual:", manual_dot)
print("u @ v:", numpy_dot)
print("np.dot:", np.dot(u, v))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">* با @ فرق دارد</h3>
<p><code>u * v</code> سه حاصل‌ضرب عنصر‌به‌عنصر می‌دهد، اما <code>u @ v</code> آن‌ها را جمع می‌کند و یک عدد می‌دهد.</p>
</div>
</div>

```python
print("element-wise:", u * v)
print("dot product:", u @ v)
```

<h2 style="text-align:right;color:inherit;">طول بردار و فاصله</h2>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>طول یا Norm بردار دوبعدی <code>[x,y]</code> از رابطهٔ فیثاغورس به‌دست می‌آید. NumPy این محاسبه را با <code>np.linalg.norm</code> انجام می‌دهد.</p>
</div>

```python
p = np.array([3, 4])

print("length:", np.linalg.norm(p))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>فاصلهٔ دو نقطه برابر طول بردار اختلاف آن‌هاست: <code>distance(a,b) = norm(a-b)</code>.</p>
</div>

```python
point_a = np.array([1, 2])
point_b = np.array([4, 6])

distance = np.linalg.norm(point_a - point_b)
print("distance:", distance)
```

<h2 style="text-align:right;color:inherit;">ضرب ماتریس در بردار</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>هر سطر ماتریس را می‌توان یک نمونه و هر ستون را یک ویژگی دانست. ضرب <code>X @ w</code> برای هر سطر یک Dot Product جداگانه حساب می‌کند.</p>
</div>

```python
X = np.array([
    [18, 16, 20],
    [14, 19, 17],
    [20, 18, 19],
    [15, 17, 16]
])

w = np.array([0.2, 0.3, 0.5])
weighted_scores = X @ w

print("X shape:", X.shape)
print("w shape:", w.shape)
print("result shape:", weighted_scores.shape)
print("weighted scores:", weighted_scores)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">قانون سازگاری شکل‌ها</h3>
<p>در ضرب <code>(4,3) @ (3,)</code> دو عدد ۳ باید هماهنگ باشند. خروجی برای چهار سطر، چهار مقدار دارد و شکل آن <code>(4,)</code> است.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">یک نورون بسیار ساده</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>محاسبهٔ پایهٔ یک نورون مصنوعی، جمع وزن‌دار ورودی‌ها به‌اضافهٔ یک Bias است: <code>z = x @ w + b</code>. سپس با یک شرط ساده می‌توان خروجی صفر یا یک ساخت.</p>
</div>

```python
X = np.array([
    [0.8, 0.6, 1.0],
    [0.2, 0.4, 0.3],
    [0.9, 0.7, 0.8],
    [0.3, 0.8, 0.5]
])

w = np.array([0.6, 0.3, 0.5])
bias = -0.7

z = X @ w + bias
prediction = (z >= 0).astype(int)

print("z:", z)
print("prediction:", prediction)
```

<h3 style="text-align:right;color:inherit;margin-top:0;">این هنوز آموزش شبکهٔ عصبی نیست</h3>
<p>در این مثال وزن‌ها از قبل داده شده‌اند. در شبکهٔ عصبی واقعی، الگوریتم باید وزن‌ها و Bias مناسب را از داده‌ها یاد بگیرد.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">ضرب ماتریس در ماتریس</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>در <code>A @ B</code> هر خانهٔ خروجی، Dot Product یک سطر از A و یک ستون از B است. تعداد ستون‌های A باید با تعداد سطرهای B برابر باشد.</p>
</div>

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [1, 2],
    [0, 1],
    [2, 0]
])

C = A @ B

print("A shape:", A.shape)
print("B shape:", B.shape)
print("C:")
print(C)
print("C shape:", C.shape)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">شکل خروجی ضرب ماتریسی</h3>
<p><code>(2,3) @ (3,2) → (2,2)</code>. ابعاد داخلی ۳ با هم حذف می‌شوند و ابعاد بیرونی ۲ و ۲ شکل خروجی را می‌سازند.</p>
</div>
</div>

```python
print("A * A:")
print(A * A)

print("A @ A.T:")
print(A @ A.T)
```
