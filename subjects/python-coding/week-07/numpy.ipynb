<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>آرایهٔ زیر نمره‌های سه دانش‌آموز در سه آزمون است. پیش از اجرا، <code>shape</code>، <code>ndim</code> و <code>size</code> را حدس بزنید.</p>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>Index</code> نشانی یک عنصر است. مانند List، شماره‌گذاری از صفر شروع می‌شود؛ بنابراین عنصر اول Index صفر دارد. Indexهای منفی نیز از انتهای Array می‌شمارند.</p>

<h2 style="text-align:right;color:inherit;">Slicing در Array یک‌بعدی</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>با الگوی <code>array[start:stop:step]</code> می‌توانیم چند عضو پیوسته یا بافاصله را انتخاب کنیم. مقدار <code>start</code> داخل برش است، اما <code>stop</code> داخل برش نیست.</p>
</div>

```python
a = np.arange(10)

print("a:", a)
print("a[2:6]:", a[2:6])
print("a[:4]:", a[:4])
print("a[6:]:", a[6:])
print("a[::2]:", a[::2])
print("a[::-1]:", a[::-1])
```

<h2 style="text-align:right;color:inherit;">Indexing و Slicing در Array دوبعدی</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>در Array دوبعدی، نشانی کامل یک عنصر به شکل <code>array[row, column]</code> است. عدد اول سطر و عدد دوم ستون را مشخص می‌کند.</p>
</div>

```python
scores = np.array([
    [18, 16, 20],
    [14, 19, 17],
    [20, 20, 18]
])

print(scores)
```

```python
print("row 0, column 1:", scores[0, 1])
print("row 2, column 0:", scores[2, 0])
print("last row, last column:", scores[-1, -1])
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>برای انتخاب یک سطر، فقط Index همان سطر را می‌نویسیم. برای انتخاب یک ستون، جای سطرها علامت <code>:</code> می‌گذاریم؛ یعنی «همهٔ سطرها».</p>
</div>

```python
print("second student:", scores[1])
print("first test:", scores[:, 0])
print("last test:", scores[:, -1])
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>می‌توانیم در هر دو جهت Slice بزنیم. عبارت زیر سطرهای اول و دوم و ستون‌های دوم و سوم را انتخاب می‌کند.</p>
</div>

```python
part = scores[0:2, 1:3]

print(part)
print("shape:", part.shape)
```

<h2 style="text-align:right;color:inherit;">Boolean Masking</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>یک شرط روی Array، یک Array از مقدارهای <code>True</code> و <code>False</code> می‌سازد. اگر همین شرط را داخل کروشه بگذاریم، فقط داده‌هایی که شرطشان درست است باقی می‌مانند.</p>
</div>

```python
scores_1d = np.array([12, 18, 20, 15, 9])

mask = scores_1d > 15

print("mask:", mask)
print("selected:", scores_1d[mask])
```

<h2 style="text-align:right;color:inherit;">Aggregation و مفهوم axis</h2>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>تابع‌های تجمیعی چند مقدار را به یک نتیجه خلاصه می‌کنند. نمونه‌های مهم این جلسه <code>sum</code>، <code>mean</code>، <code>min</code> و <code>max</code> هستند.</p>

```python
print("mean of all scores:", scores.mean())
print("sum of all scores:", scores.sum())
print("minimum:", scores.min())
print("maximum:", scores.max())
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>در جدول نمره‌ها، <code>axis=0</code> سطرها را جمع می‌کند و برای هر <strong>ستون</strong> یک جواب می‌دهد. <code>axis=1</code> ستون‌ها را جمع می‌کند و برای هر <strong>سطر</strong> یک جواب می‌دهد.</p>

```python
print("mean of each test:", scores.mean(axis=0))
print("mean of each student:", scores.mean(axis=1))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">تغییر چیدمان با reshape و transpose</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>reshape</code> داده‌ها را عوض نمی‌کند؛ فقط آن‌ها را با شکل تازه می‌چیند. تعداد کل عناصر شکل قدیم و جدید باید برابر باشد.</p>
</div>

```python
a = np.arange(12)
b = a.reshape(3, 4)

print("a:", a)
print("a.shape:", a.shape)
print("b:\n", b)
print("b.shape:", b.shape)
```

<h3 style="text-align:right;color:inherit;margin-top:0;">قانون reshape</h3>
<p>آرایهٔ ۱۲ عضوی می‌تواند شکل‌های <code>(2, 6)</code>، <code>(3, 4)</code> یا <code>(4, 3)</code> داشته باشد، اما شکل <code>(5, 3)</code> ممکن نیست؛ زیرا به ۱۵ عنصر نیاز دارد.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>transpose</code> یا ویژگی <code>.T</code> جای سطرها و ستون‌ها را عوض می‌کند.</p>
</div>

```python
print("original:\n", b)
print("transposed:\n", b.T)
print("original shape:", b.shape)
print("transposed shape:", b.T.shape)
```
