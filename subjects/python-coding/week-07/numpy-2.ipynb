<h3 style="text-align:right;color:inherit;margin-top:0;">هدف این ساعت</h3>
<p>قرار است از «انتخاب داده» یک قدم جلوتر برویم و به سؤال‌های واقعی‌تری پاسخ دهیم:</p>
<ul><li>چطور داده‌ها را مرتب کنیم و ترتیب واقعی آن‌ها را نگه داریم؟</li><li>چطور جای بیشترین و کمترین مقدار را پیدا کنیم؟</li><li>چطور با <code>Broadcasting</code> روی هر سطر یا ستون مقدار متفاوتی اعمال کنیم؟</li><li>چطور چند Array را از جهت سطر یا ستون به هم وصل کنیم؟</li></ul>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">دادهٔ محوری جلسه</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>هر سطر یک دانش‌آموز و هر ستون یک آزمون است. نام‌ها نیز با ترتیب سطرها هماهنگ‌اند.</p>
</div>

```python
import numpy as np

names = np.array(["Ali", "Nima", "Reza", "Ahmad"])
scores = np.array([
    [18, 15, 20],
    [16, 19, 17],
    [20, 18, 19],
    [14, 17, 16]
])

print(names)
print(scores)
print("shape:", scores.shape)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">مرتب‌سازی با sort</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>np.sort</code> مقدارها را مرتب می‌کند و یک Array جدید برمی‌گرداند. Array اصلی تغییر نمی‌کند.</p>
</div>

```python
values = np.array([12, 7, 20, 15, 9])

ascending = np.sort(values)
descending = np.sort(values)[::-1]

print("original:", values)
print("ascending:", ascending)
print("descending:", descending)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>در Array دوبعدی، <code>axis=1</code> هر سطر و <code>axis=0</code> هر ستون را جداگانه مرتب می‌کند.</p>
</div>

```python
print("sort each row:\n", np.sort(scores, axis=1))
print("sort each column:\n", np.sort(scores, axis=0))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">نکتهٔ مهم sort</h3>
<p>مرتب‌کردن خود نمره‌ها ارتباط آن‌ها را با نام دانش‌آموز از بین می‌برد. برای رتبه‌بندی افراد به‌جای <code>sort</code> به Indexهای ترتیب نیاز داریم.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">حفظ ترتیب با argsort</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>argsort</code> به‌جای مقدارهای مرتب‌شده، Indexهایی را می‌دهد که داده‌ها را مرتب می‌کنند.</p>
</div>

```python
averages = scores.mean(axis=1)
order = np.argsort(averages)[::-1]

print("averages:", averages)
print("order:", order)
print("ranked names:", names[order])
print("ranked averages:", averages[order])
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">چرا argsort؟</h3>
<p>اگر فقط میانگین‌ها را مرتب کنیم، دیگر نمی‌دانیم هر میانگین متعلق به چه کسی است. Indexهای <code>argsort</code> را هم‌زمان روی <code>names</code> و <code>averages</code> اعمال می‌کنیم.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><strong>تمرین:</strong> نمره‌های آزمون دوم را از بیشترین به کمترین مرتب کنید و نام دانش‌آموزان را نیز با همان ترتیب چاپ کنید.</p>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">جای بیشینه و کمینه با argmax و argmin</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>max</code> خود مقدار بیشینه را می‌دهد؛ <code>argmax</code> Index آن را برمی‌گرداند. همین تفاوت برای <code>min</code> و <code>argmin</code> وجود دارد.</p>
</div>

```python
print("best average:", averages.max())
print("best student index:", averages.argmax())
print("best student name:", names[averages.argmax()])

print("lowest average:", averages.min())
print("student name:", names[averages.argmin()])
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>با تعیین <code>axis</code> می‌توانیم جای بیشترین مقدار هر سطر یا ستون را پیدا کنیم.</p>
</div>

```python
print("best student index in each exam:", scores.argmax(axis=0))
print("best exam index for each student:", scores.argmax(axis=1))

best_in_each_exam = scores.argmax(axis=0)
print("best names in each exam:", names[best_in_each_exam])
```

<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #2563eb;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">شمارهٔ انسانی و Index</h3>
<p>خروجی <code>argmax</code> یک Index صفرمبنا است. اگر فقط شمارهٔ آزمون برای نمایش به انسان لازم باشد، می‌توانیم یک واحد به آن اضافه کنیم.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">Broadcasting: عملیات بدون ساختن جدول تکراری</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>قبلاً یک عدد را به کل Array اضافه کردیم. Broadcasting اجازه می‌دهد یک بردار سازگار نیز خودکار روی سطرها یا ستون‌ها تکرار شود.</p>
</div>

```python
exam_bonus = np.array([1, 0, 2])
adjusted_scores = scores + exam_bonus

print("exam bonus:", exam_bonus)
print("adjusted scores:\n", adjusted_scores)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>بردار سه‌عضوی روی هر سطر اعمال شد؛ چون جدول سه ستون دارد. برای افزودن مقدار متفاوت به هر دانش‌آموز، بردار را ستونی می‌کنیم.</p>
</div>

```python
student_bonus = np.array([1, 0, -1, 2]).reshape(4, 1)
adjusted_by_student = scores + student_bonus

print("student bonus shape:", student_bonus.shape)
print(adjusted_by_student)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<div style="border:1px solid rgba(148,163,184,0.35);border-right:5px solid #0ea5e9;background:rgba(148,163,184,0.08);padding:10px 14px;border-radius:8px;">
<h3 style="text-align:right;color:inherit;margin-top:0;">قانون سادهٔ Broadcasting</h3>
<p>ابعاد را از سمت راست مقایسه کنید. هر دو اندازه باید برابر باشند یا یکی از آن‌ها <code>1</code> باشد. شکل <code>(4, 3)</code> با <code>(3,)</code> و <code>(4, 1)</code> سازگار است، اما با <code>(4,)</code> سازگار نیست.</p>
</div>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>یک کاربرد مهم در Machine Learning، کم‌کردن میانگین هر ویژگی یا ستون از داده‌هاست.</p>
</div>

```python
exam_means = scores.mean(axis=0)
centered_scores = scores - exam_means

print("exam means:", exam_means)
print("centered scores:\n", centered_scores)
print("new column means:", centered_scores.mean(axis=0))
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<h2 style="text-align:right;color:inherit;">ترکیب Arrayها</h2>
</div>

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p>برای افزودن سطر جدید از <code>vstack</code> و برای افزودن ستون جدید از <code>hstack</code> استفاده می‌کنیم.</p>
</div>

```python
new_student = np.array([[17, 20, 18]])
scores_with_student = np.vstack([scores, new_student])

print(scores_with_student)
print("shape:", scores_with_student.shape)
```

```python
project_score = np.array([[19], [18], [20], [17]])
scores_with_project = np.hstack([scores, project_score])

print(scores_with_project)
print("shape:", scores_with_project.shape)
```

<div dir="rtl" lang="fa" style="direction:rtl;text-align:right;line-height:2;color:inherit;">
<p><code>concatenate</code> شکل عمومی‌تر همین اتصال است. <code>axis=0</code> سطر و <code>axis=1</code> ستون اضافه می‌کند.</p>
</div>

```python
same_rows = np.concatenate([scores, new_student], axis=0)
same_columns = np.concatenate([scores, project_score], axis=1)

print(same_rows.shape)
print(same_columns.shape)
```
